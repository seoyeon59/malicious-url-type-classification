```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
import warnings
warnings.filterwarnings("ignore")
```

```python
df = pd.read_csv('all.csv')
```

```python
df.info()
```

```python
df.columns
```

```python
select_col = [
     'SymbolCount_Domain', 'SymbolCount_Directoryname',
       'SymbolCount_FileName', 'SymbolCount_Extension',
       'SymbolCount_Afterpath', 'URL_Type_obf_Type'
]
```

```python
select_df = df[select_col]
```

```python
# -1 값의 개수를 컬럼별로 계산
(select_df == -1).sum()
```

```python
select_df.isnull().sum()
```

```python
# -1이 있는 행 삭제
df_cleaned = select_df[~(select_df == -1).any(axis=1)]
```

```python
# -1 값의 개수를 컬럼별로 계산
(df_cleaned == -1).sum()
```

```python
df_cleaned.info()
```

```python
df_cleaned['URL_Type_obf_Type'].value_counts()
```

```python
df_cleaned = df_cleaned[df_cleaned['URL_Type_obf_Type'] != 'benign']
```

```python
df_cleaned['URL_Type_obf_Type'].value_counts()
```

```python
df_cleaned.columns
```

```python
X_col = [
   'SymbolCount_Domain', 'SymbolCount_Directoryname',
       'SymbolCount_FileName', 'SymbolCount_Extension',
       'SymbolCount_Afterpath'
]
```

```python
X = df_cleaned[X_col]
y = df_cleaned['URL_Type_obf_Type']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import classification_report, confusion_matrix, roc_auc_score, RocCurveDisplay
```

```python
scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
model = DecisionTreeClassifier(
    random_state=42
)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]
```

```python
print("📌 [Confusion Matrix]")
print(confusion_matrix(y_test, y_pred))
print("\n📌 [Classification Report]")
print(classification_report(y_test, y_pred))
```

```python
model = DecisionTreeClassifier(
    random_state=42,
    max_depth = 20
)
model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
y_prob = model.predict_proba(X_test)[:, 1]
```

```python
print("📌 [Confusion Matrix]")
print(confusion_matrix(y_test, y_pred))
print("\n📌 [Classification Report]")
print(classification_report(y_test, y_pred))
```
